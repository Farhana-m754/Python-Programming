```python
import pandas as pd
```

```python
print(marks.iloc[0])
```

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
print(marks.iloc[2])
```

```python
marks = marks.drop('Rank5')

print(marks)
```
